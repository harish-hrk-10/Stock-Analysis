```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
df = pd.read_excel("/content/shopify_stock.xlsx")

# Basic information
print(df.shape)
print(df.head())
df.info()
print(df.describe())

# Convert date
df["date"] = pd.to_datetime(df["date"], utc=True)

# Check missing values
print(df.isnull().sum())

# Remove missing values
df = df.dropna()

# Check and remove duplicates
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()

# Sort by date
df = df.sort_values("date")

# Calculate daily changes
df["Daily_Delta"] = df["close"] - df["open"]

df["Daily_return"] = (
    (df["close"] - df["open"]) / df["open"]
) * 100

# Display calculations
print(
    df[
        ["date", "open", "close", "Daily_Delta", "Daily_return"]
    ].head()
)

# Volume statistics
avg_volume = df["volume"].mean()
max_volume = df["volume"].max()
min_volume = df["volume"].min()

print("Average_volume:", avg_volume)
print("Maximum_volume:", max_volume)
print("Minimum_volume:", min_volume)

# Define anomaly threshold
threshold = avg_volume + (2 * df["volume"].std())

print("Anomaly Threshold:", threshold)

# Find anomalous trading days
anomalous_days = df[df["volume"] > threshold]

print("\nAnomalous Trading Days:")

print(
    anomalous_days.head()[
        [
            "date",
            "open",
            "close",
            "volume",
            "Daily_Delta",
            "Daily_return"
        ]
    ]
)

# Return statistics
print("Mean Return:", df["Daily_return"].mean())
print("Variance:", df["Daily_return"].var())
print("Standard Deviation:", df["Daily_return"].std())
```
